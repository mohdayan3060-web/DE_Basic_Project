 API fetching & pagination logic

```python
import pandas as pd
import requests
```

```python
response=requests.get("https://api.freeapi.app/api/v1/public/stocks")
response.json()
```

```python
pd.DataFrame(response.json()['data']['data'])
```

```python
all_stock=[]
for i in range(192):
    response=requests.get("https://api.freeapi.app/api/v1/public/stocks",params={
        'page':i})
    if response.status_code==200:
        page_data=response.json()['data']['data']
        all_stock.extend(page_data)
    else:
        print(f"failed fetch page{i}")    
   
df_stock=pd.DataFrame(all_stock)
df_stock
```

```python
df_stock.info()
df_stock.head()
```

DATA CLEANING 

```python
df_stock.columns

#droping ISIN columns 
df_stock.drop(columns='ISIN', inplace=True )

```

```python
#columns Renaming  into snake_case
df_stock = df_stock.rename(columns={
    'Name': 'name',
    'Symbol': 'symbol',
    'ListingDate': 'listing_date',
    'MarketCap': 'market_cap',
    'CurrentPrice': 'current_price',
    'HighLow': 'high_low',
    'StockPE': 'stock_pe',
    'BookValue':'book_value',
    'DividendYield':'dividend_yield',
    'ROCE':'roce',
    'ROE':'roe',
    'FaceValue':'face_value',
})


```

```python
# using list comprehention  exract symbol only 
df_stock.sample(3)
df_stock['symbol']=[symbol.strip('0123456789') for symbol in df_stock['symbol']]
df_stock.info()
```

```python
# stripe -> space , symbole and comma  and covert into float

df_stock
df_stock['current_price']=df_stock.current_price.str.replace(",","").str.replace("₹","")
# converting into float and also dealing with empty string 
df_stock['current_price']=pd.to_numeric(df_stock['current_price'],errors='coerce').fillna(0.0)


df_stock['book_value']=df_stock.book_value.str.replace(",","").str.replace("₹","")
df_stock['book_value']=pd.to_numeric(df_stock['book_value'],errors='coerce').fillna(0.0)

df_stock['face_value']=df_stock.face_value.str.replace(",","").str.replace("₹","")
df_stock['face_value']=pd.to_numeric(df_stock['face_value'],errors='coerce').fillna(0.0)



```
